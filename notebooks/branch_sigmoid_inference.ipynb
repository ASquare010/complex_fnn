Small language-model inference; 256-token context.

```python
from pathlib import Path
import os, sys
root = Path.cwd()
if not (root / 'src').exists(): root = root.parent
assert (root / 'src/models/branch_sigmoid').is_dir()
os.chdir(root)
sys.path.insert(0, str(root / 'src'))
```

```python
import torch
from models.branch_sigmoid.loader import load_winner
model, tokenizer = load_winner('cpu')
text = 'Once upon a time a'
ids = tokenizer.encode(text).ids
with torch.no_grad():
    for _ in range(32):
        tokens = torch.tensor([ids[-model.config.context:]])
        ids.append(int(model(tokens)[0, -1].argmax()))
print(tokenizer.decode(ids))
```
